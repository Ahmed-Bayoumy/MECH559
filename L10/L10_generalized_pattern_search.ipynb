# Generalized Pattern Search (GPS)
### MECH 559 - Systems Optimization - L10
### Course Instructor: Dr. Ahmed BayoumyAuthor of the notebook: Dr. Ahmed Bayoumy
Companion notebook to the *Search methods* section of the lecture. GPS (Torczon, 1997)
generalizes coordinate search and Hooke & Jeeves: every trial point lies on a **mesh**
$M_k = \{x + \Delta_k D z : z \in \mathbb{N}^{n_D}\}$, and each iteration has an optional global
**Search** step and a local **Poll** step along the columns of a **positive spanning set**
$D$. We implement GPS, compare direction sets and mesh-update rules, check its convergence
behaviour on a smooth function, and show the limitation that motivates MADS.

## Learning objectives

By the end, students should be able to:

1. Verify that a direction set $D$ positively spans $\mathbb{R}^n$, and compare the coordinate
   set $[I_n\ -I_n]$ ($2n$ directions) with a minimal positive basis ($n+1$ directions).
2. Implement GPS with Search and Poll steps on the mesh and the mesh-size update
   (possibly increase $\Delta_k$ on success, decrease it on failure).
3. Interpret the GPS convergence result: on a smooth function the limit is stationary; on a
   nonsmooth function only the directions in $D$ are guaranteed to be non-descent.
4. Explain why **any** fixed finite $D$ can be defeated by a nonsmooth function.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    import ipywidgets as widgets
    from IPython.display import display
    WIDGETS_AVAILABLE = True
except ImportError:
    WIDGETS_AVAILABLE = False

plt.rcParams.update({"font.size": 11, "figure.dpi": 110, "axes.grid": True, "grid.alpha": 0.25})
COLORS = {"coord": "#1D5FA8", "minimal": "#D1495B", "search": "#2A9D5C", "mesh": "#BBBBBB",
          "poll": "#D1495B", "optimum": "#EDAE49", "incumbent": "#264653"}
print(f"Optional widgets available: {WIDGETS_AVAILABLE}")
```

## 1. Positive spanning sets
#
$D$ **positively spans** $\mathbb{R}^n$ if every vector is a nonnegative combination of its
columns. Equivalently, for every nonzero $w$ some column satisfies $d^\top w > 0$: whatever the
(unknown) gradient, at least one poll direction makes an acute angle with $-\nabla f$, i.e.
is a descent direction for smooth $f$.

```python
D_coord = np.array([[1, 0], [0, 1], [-1, 0], [0, -1]], dtype=float)   # [I -I], 2n directions
D_min = np.array([[1, 0], [0, 1], [-1, -1]], dtype=float)              # [I -1], n+1 directions
D_bad = np.array([[1, 0], [0, 1], [-1, 0]], dtype=float)               # NOT positively spanning

angles = np.deg2rad(np.arange(0.0, 360.0, 0.1))           # unit vectors w every 0.1 degree
W = np.column_stack([np.cos(angles), np.sin(angles)])
for name, D in [("coordinate [I -I]", D_coord), ("minimal [I -1]", D_min), ("[e1, e2, -e1]", D_bad)]:
    worst_cos = np.min(np.max(W @ (D / np.linalg.norm(D, axis=1, keepdims=True)).T, axis=1))
    print(f"{name:18s}: min over w of max_d cos(d, w) = {worst_cos:+.3f}  "
          f"-> {'positive spanning' if worst_cos > 1e-9 else 'NOT positive spanning'}")
```

The worst-case cosine measures how well $D$ "covers" all directions: $1/\sqrt2 \approx 0.707$
for the coordinate set, $\approx 0.38$ for the minimal basis (fewer directions, worse
coverage), and $0$ for $[e_1, e_2, -e_1]$: for $w = (0, -1)$ no column makes an acute angle,
so this set can never move "down" - it is not a positive spanning set.

## 2. The GPS algorithm
#
From the slides' general framework:
#
- **[1.1] Search** (optional): evaluate a finite number of mesh points chosen by any rule.
  Here a speculative step that repeats the last successful direction with twice the step.
- **[1.2] Poll** (if the Search failed): $P_k = \{x_k + \Delta_k d : d \in D\}$, evaluated
  opportunistically.
- **[2] Update**: on success $x_{k+1}$ is the improving point and $\Delta$ may be increased
  ($\Delta_{k+1} = \tau\Delta_k$, $\tau \ge 1$); on failure $\Delta_{k+1} = \Delta_k / 2$.

```python
def gps(func, x0, D, delta0=1.0, tau=2.0, search=False, budget=1000, tol=1e-9):
    x, fx, delta = np.array(x0, dtype=float), func(x0), delta0
    n_eval, last_dir = 1, None
    hist = [{"x": x.copy(), "f": fx, "delta": delta, "n_eval": n_eval, "poll": []}]
    while n_eval < budget and delta > tol:
        success, poll_pts = False, []
        if search and last_dir is not None:                    # [1.1] Search
            t = x + 2 * delta * last_dir
            ft = func(t); n_eval += 1
            if ft < fx:
                x, fx, success = t, ft, True
        if not success:                                         # [1.2] Poll
            for d in D:
                t = x + delta * d
                ft = func(t); n_eval += 1
                poll_pts.append(t)
                if ft < fx:
                    x, fx, success, last_dir = t, ft, True, d
                    break
        if success:                                             # [2] Update
            delta *= tau
        else:
            delta /= 2.0
            last_dir = None
        hist.append({"x": x.copy(), "f": fx, "delta": delta, "n_eval": n_eval, "poll": poll_pts})
    return x, fx, hist


def f(x):
    return (5 * x[0] - 2) ** 4 + (5 * x[0] - 2) ** 2 * x[1] ** 2 + (3 * x[1] + 1) ** 2


def grad_f(x):
    a = 5 * x[0] - 2
    return np.array([20 * a ** 3 + 10 * a * x[1] ** 2, 2 * a ** 2 * x[1] + 6 * (3 * x[1] + 1)])


x0 = np.array([2.0, 2.0])
x_star = np.array([0.4, -1.0 / 3.0])
```

---
## 3. Comparing direction sets and update rules
#
Same test function as in the coordinate search notebook. We count the evaluations needed to
reach $f < 10^{-6}$.
#
> **Predict:** which helps more here - fewer poll directions, or expanding $\Delta$ after a
> success?

```python
def evals_to_reach(target, **kwargs):
    _, _, hist = gps(f, x0, budget=3000, **kwargs)
    return next((h["n_eval"] for h in hist if h["f"] < target), np.nan)


variants = {
    "CS: coordinate D, tau = 1": dict(D=D_coord, tau=1.0),
    "GPS: coordinate D, tau = 2": dict(D=D_coord, tau=2.0),
    "GPS: minimal D, tau = 1": dict(D=D_min, tau=1.0),
    "GPS: minimal D, tau = 2": dict(D=D_min, tau=2.0),
    "GPS: coordinate D, tau = 2 + Search": dict(D=D_coord, tau=2.0, search=True),
    "GPS: minimal D, tau = 2 + Search": dict(D=D_min, tau=2.0, search=True),
}
table = pd.DataFrame({"evaluations to reach f < 1e-6": {k: evals_to_reach(1e-6, **v) for k, v in variants.items()}})
table
```

On this problem the **minimal basis** is cheapest (3 poll points instead of 4), while
expanding the mesh after a success and our speculative Search step both *cost* evaluations:
the incumbent overshoots and the mesh has to be refined again. This is problem-dependent -
expansion and good Search steps (e.g. surrogate-based, see L11) pay off when the start is far
from the solution - which is why GPS leaves these choices to the user.

## 4. Mesh refinement and the poll set

```python
_, _, hist = gps(f, x0, D_coord, tau=1.0, budget=400)
failures = [k for k, h in enumerate(hist[1:], start=1) if h["delta"] < hist[k - 1]["delta"]]
fig, axes = plt.subplots(1, 3, figsize=(14, 4.4))
for ax, k in zip(axes, failures[2:5]):
    h_prev, h = hist[k - 1], hist[k]
    d_k, xk = h_prev["delta"], h_prev["x"]
    lo, hi = xk - 2.2 * d_k, xk + 2.2 * d_k
    ax.grid(False)
    for spacing, style in [(d_k, "-"), (d_k / 2, ":")]:      # current mesh and the refined one
        for v in xk[0] + spacing * np.arange(-6, 7):
            ax.axvline(v, color=COLORS["mesh"], lw=0.9, ls=style)
        for v in xk[1] + spacing * np.arange(-6, 7):
            ax.axhline(v, color=COLORS["mesh"], lw=0.9, ls=style)
    for p in h["poll"]:
        ax.plot([xk[0], p[0]], [xk[1], p[1]], color=COLORS["poll"], lw=1.5)
        ax.plot(*p, "o", color=COLORS["poll"], ms=7)
    ax.plot(*xk, "s", color=COLORS["incumbent"], ms=9, label="$x_k$")
    ax.set(xlim=(lo[0], hi[0]), ylim=(lo[1], hi[1]), aspect="equal",
           title=f"iteration {k}: $\\Delta_k$ = {d_k:g}, poll fails")
    ax.legend(loc="upper right", fontsize=8)
plt.tight_layout()
plt.show()
```

Each failed poll means $x_k$ is a **mesh local optimizer**: it is better than all its poll
neighbours on the current mesh (solid lines). The mesh is then refined to $\Delta_k / 2$
(dotted lines) and polling resumes on the finer mesh.

## 5. What GPS guarantees, and what it does not
#
The GPS convergence result (Clarke calculus, slide *Coordinate Search evolution*): if
$\hat x$ is a limit of mesh local optimizers on meshes that get infinitely fine and $f$ is
Lipschitz near $\hat x$, then $f^\circ(\hat x; d) \ge 0$ for every direction $d$ used
infinitely often. For smooth $f$ the directions positively span $\mathbb{R}^n$, so
$\nabla f(\hat x) = 0$.

```python
x_end, f_end, hist = gps(f, x0, D_min, tau=1.0, budget=2000, tol=1e-12)
print(f"GPS (minimal D): x = {np.round(x_end, 8)}, f = {f_end:.1e}, ||grad f|| = {np.linalg.norm(grad_f(x_end)):.1e}")
```

# On a **nonsmooth** function the guarantee only covers the directions in $D$. Reuse the
# kink function of the coordinate search notebook, $g(x) = |x_1 - x_2| + \tfrac12|x_1 + x_2|$,
# started on its kink at $(1, 1)$, and a rotated copy
# $g_2(x) = |x_1 + x_2| + \tfrac12|x_1 - x_2|$ started on its kink at $(1, -1)$. Both are
# minimized at the origin.

```python
def g(x):
    return abs(x[0] - x[1]) + 0.5 * abs(x[0] + x[1])


def g2(x):
    return abs(x[0] + x[1]) + 0.5 * abs(x[0] - x[1])


rows = {}
for fname, func, start in [("g from (1, 1)", g, (1.0, 1.0)), ("g2 from (1, -1)", g2, (1.0, -1.0))]:
    for dname, D in [("coordinate D", D_coord), ("minimal D", D_min)]:
        xe, fe, _ = gps(func, start, D, tau=1.0, budget=2000)
        rows[(fname, dname)] = {"final x": tuple(np.round(xe, 6)), "final value": round(fe, 6)}
pd.DataFrame(rows).T
```

# The minimal basis contains $-(1,1)$, which lies inside the descent cone of $g$, so it solves
# $g$ - but on the rotated $g_2$ none of its three directions is a descent direction, and it
# stalls just like the coordinate set. For **every** finite $D$ one can rotate a kink so that
# the descent cone falls between the directions. MADS removes this weakness by changing the
# poll directions at every iteration so that, asymptotically, they become dense in the unit
# sphere (next notebook).

```python
def explore_gps(tau=1.0, use_min_basis=True, budget=300):
    D = D_min if use_min_basis else D_coord
    _, _, h_ = gps(f, x0, D, tau=tau, budget=budget)
    path = np.array([h["x"] for h in h_])
    xs = np.linspace(-0.5, 2.5, 300)
    ys = np.linspace(-1.0, 2.5, 300)
    X, Y = np.meshgrid(xs, ys)
    fig, ax = plt.subplots(figsize=(6, 4.8))
    ax.contour(X, Y, np.log10(f([X, Y]) + 1e-12), levels=25, cmap="Greys", linewidths=0.6)
    ax.plot(path[:, 0], path[:, 1], "o-", ms=3, color=COLORS["minimal" if use_min_basis else "coord"])
    ax.plot(*x_star, "*", color=COLORS["optimum"], ms=16, mec="k")
    ax.set(xlabel="$x_1$", ylabel="$x_2$",
           title=f"{'minimal' if use_min_basis else 'coordinate'} D, tau={tau:g}: f = {h_[-1]['f']:.1e}")
    plt.show()


if WIDGETS_AVAILABLE:
    display(widgets.interactive(
        explore_gps,
        tau=widgets.FloatSlider(value=1.0, min=1.0, max=4.0, step=0.5, description="tau"),
        use_min_basis=widgets.Checkbox(value=True, description="minimal basis"),
        budget=widgets.IntSlider(value=300, min=20, max=1000, step=20),
    ))
else:
    explore_gps(1.0, True, 300)
```

---
## Takeaways
#
1. GPS polls along a positive spanning set $D$ on a mesh $M_k$; a positive spanning set
   always contains a descent direction of a smooth function.
2. The Search step, the number of directions and the mesh expansion factor are free design
   choices; their value is problem-dependent (here: minimal basis, no expansion was best).
3. On smooth functions GPS limit points are stationary.
4. On nonsmooth functions GPS only guarantees $f^\circ(\hat x; d) \ge 0$ for $d \in D$; a kink
   whose descent cone falls between the directions of $D$ stops it at a non-minimizer.
   MADS fixes this with asymptotically dense directions.
