# Nelder-Mead on Noisy Blackboxes
### MECH 559 - Systems Optimization - L10

Companion notebook to the "Nelder-Mead on noisy blackboxes" slide. Physical experiments and
many simulations return a **noisy** value: calling the blackbox twice at the same $x$ gives
two different answers. NM only compares function values, so noise directly corrupts every
decision it makes. We reproduce the slide's experiment, separate what is real from what is
not in a frequently quoted claim ("noise makes NM converge to a maximizer"), and test simple
remedies.

## Learning objectives

By the end, students should be able to:

1. Explain why NM **stagnates** when the true variation of $f$ across the simplex is smaller
   than the noise level.
2. Explain why the value NM reports at the end is **optimistically biased** (it keeps its
   luckiest evaluation).
3. Choose the initial simplex size relative to the noise, and use re-evaluation/averaging to
   trade evaluations for accuracy.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

try:
    import ipywidgets as widgets
    from IPython.display import display
    WIDGETS_AVAILABLE = True
except ImportError:
    WIDGETS_AVAILABLE = False

plt.rcParams.update({"font.size": 11, "figure.dpi": 110, "axes.grid": True, "grid.alpha": 0.25})
COLORS = {"peak": "#D1495B", "slope1": "#1D5FA8", "slope2": "#2A9D5C", "true": "#264653", "reported": "#EDAE49"}
print(f"Optional widgets available: {WIDGETS_AVAILABLE}")
```

---
## 1. A noisy blackbox
#
We **minimize** $f(x) = 10\,e^{-\|x\|^2}$, a peak at the origin, through the blackbox
$f(x) + \varepsilon$ with fresh noise $\varepsilon \sim \mathcal{N}(0, \sigma^2)$, $\sigma = 2$,
at every call. The minimum is "at infinity" ($f \to 0$), so a working method should move
**away** from the peak.

```python
SIGMA = 2.0


def f_true(x):
    return 10.0 * np.exp(-(x[0] ** 2 + x[1] ** 2))


def make_noisy(sigma, seed):
    rng = np.random.default_rng(seed)
    return lambda x: f_true(x) + rng.normal(0.0, sigma)


def run_nm(x0, edge=None, sigma=SIGMA, seed=0, maxfev=400, n_avg=1):
    noisy = make_noisy(sigma, seed)
    func = noisy if n_avg == 1 else (lambda x: np.mean([noisy(x) for _ in range(n_avg)]))
    x0 = np.asarray(x0, dtype=float)
    opts = {"xatol": 1e-5, "fatol": 1e-5, "maxfev": maxfev // n_avg}
    if edge is not None:
        opts["initial_simplex"] = np.array([x0, x0 + [edge, 0.0], x0 + [0.0, edge]])
    res = minimize(func, x0, method="Nelder-Mead", options=opts)
    return res


noisy = make_noisy(SIGMA, seed=1)
print("five calls at the same point x = (0.1, 0.1):", np.round([noisy((0.1, 0.1)) for _ in range(5)], 2))
print("true value:", round(f_true((0.1, 0.1)), 3))
```

---
## 2. The default simplex stagnates wherever it starts
#
Given only `x0`, SciPy builds its initial simplex by perturbing each nonzero coordinate
by 5%: from $x^0 = (0.1, 0.1)$ the edges are $0.005$ long. Across such a small simplex the
true $f$ changes by about $0.01$, while the noise has standard deviation $2$.
#
> **Predict:** started near the peak, and started on the slope, where will 100 noisy runs
> end?

```python
STARTS = {"near the peak": (0.1, 0.1), "on the slope": (1.0, 0.5), "further down": (1.2, -0.8)}
START_COLORS = [COLORS["peak"], COLORS["slope1"], COLORS["slope2"]]
N_RUNS = 100


def final_points(edge):
    return {name: np.array([run_nm(x0, edge=edge, seed=s).x for s in range(N_RUNS)]) for name, x0 in STARTS.items()}


ends_default = final_points(None)
ends_large = final_points(0.5)

grid = np.linspace(-4, 4, 300)
GX, GY = np.meshgrid(grid, grid)
fig, axes = plt.subplots(1, 2, figsize=(12, 5.2))
for ax, ends, title in [(axes[0], ends_default, "SciPy default simplex (edge ~0.005)"),
                        (axes[1], ends_large, "initial simplex with edge 0.5")]:
    ax.contour(GX, GY, 10 * np.exp(-(GX ** 2 + GY ** 2)), levels=12, cmap="Greys", linewidths=0.6)
    for (name, x0), c in zip(STARTS.items(), START_COLORS):
        ax.scatter(ends[name][:, 0], ends[name][:, 1], s=8, color=c, alpha=0.5)
        ax.plot(*x0, "*", color=c, ms=15, mec="k", label=f"start {name}")
    ax.set(xlim=(-4, 4), ylim=(-4, 4), aspect="equal", xlabel="$x_1$", ylabel="$x_2$", title=title)
axes[0].legend(fontsize=8, loc="lower left")
plt.tight_layout()
plt.show()

for name, x0 in STARTS.items():
    moved = np.median(np.linalg.norm(ends_default[name] - np.array(x0), axis=1))
    print(f"default simplex, start {name:14s}: median distance moved = {moved:.3f}")
```

With the default simplex NM **does not move**: runs started next to the peak end next to
the peak, and runs started on the slope end on the slope. It looks as if NM "converged to
the maximizer" only because it was started there. Every NM decision compares values that
differ by $\sim 0.01$ plus noise of size $2$: the decisions are coin flips.
#
With an initial simplex of edge $0.5$, the true variation across the simplex is comparable
to the noise, and NM moves downhill - away from the peak - although the noise still
scatters the end points.
#

## 3. The reported value is too optimistic
#
NM keeps the vertex with the lowest *observed* value. With noise, that is the luckiest
draw, not the best point.

```python
reported, true_at_x, success = [], [], []
for s in range(N_RUNS):
    res = run_nm(STARTS["near the peak"], seed=s)
    reported.append(res.fun)
    true_at_x.append(f_true(res.x))
    success.append(res.success)

fig, ax = plt.subplots(figsize=(7, 4))
bins = np.linspace(-4, 12, 41)
ax.hist(reported, bins=bins, color=COLORS["reported"], alpha=0.8, label="value reported by NM")
ax.hist(true_at_x, bins=bins, color=COLORS["true"], alpha=0.8, label="true $f$ at the returned point")
ax.set(xlabel="$f$", ylabel="runs", title="Default simplex, started near the peak")
ax.legend()
plt.show()
print(f"median reported value = {np.median(reported):.2f}, median true value = {np.median(true_at_x):.2f}, "
      f"success flags: {sum(success)}/{N_RUNS}")
```

The reported values are biased low by several noise standard deviations, and SciPy itself
flags the runs as unsuccessful (`success=False`: the evaluation budget was exhausted before
the tolerances were met). Never trust the final value of a noisy run: **re-evaluate** the
returned point several times.
#
---
## 4. Remedies: a larger simplex, and averaging
#
Two simple remedies, compared with the **same total budget** of 400 blackbox calls:
#
- an initial simplex that is large compared with the noise;
- averaging $m$ calls per evaluation, which divides the noise standard deviation by
  $\sqrt m$ but leaves only $400/m$ NM evaluations.
#
A run "escapes" when it ends more than 2 units from the peak ($f < 0.2$).

```python
configs = [("default simplex", None, 1), ("edge 0.5", 0.5, 1), ("edge 0.5, average m=4", 0.5, 4),
           ("edge 0.5, average m=16", 0.5, 16)]
print(f"{'configuration':24s} | {'escaped':>8s} | {'median true f':>13s}")
for name, edge, m in configs:
    xs_ = [run_nm(STARTS["near the peak"], edge=edge, seed=s, n_avg=m).x for s in range(N_RUNS)]
    tv = np.array([f_true(x) for x in xs_])
    print(f"{name:24s} | {np.mean(tv < 0.2):8.0%} | {np.median(tv):13.3f}")
```

A simplex that matches the scale of the noise is what makes NM work at all; averaging
then trades evaluations for reliability. Here averaging pays off handsomely: even with
$m = 16$ (only 25 NM evaluations) almost every run escapes, because each comparison is now
reliable. On a problem that needs many NM iterations, a large $m$ would exhaust the budget
before NM gets far - choosing $m$ is a budget decision. Methods designed for noise (e.g.
stochastic variants of MADS) build this trade-off into the algorithm.
#
---
## 5. Explore: noise level and simplex size

```python
def explore(sigma=2.0, log10_edge=-2.3):
    edge = 10.0 ** log10_edge
    ends = np.array([run_nm((0.1, 0.1), edge=edge, sigma=sigma, seed=s).x for s in range(60)])
    fig, ax = plt.subplots(figsize=(5.5, 5))
    ax.contour(GX, GY, 10 * np.exp(-(GX ** 2 + GY ** 2)), levels=12, cmap="Greys", linewidths=0.6)
    ax.scatter(ends[:, 0], ends[:, 1], s=10, color=COLORS["peak"], alpha=0.6)
    ax.plot(0.1, 0.1, "*", color=COLORS["peak"], ms=15, mec="k")
    escaped = np.mean(np.linalg.norm(ends, axis=1) > 2)
    ax.set(xlim=(-4, 4), ylim=(-4, 4), aspect="equal",
           title=f"sigma = {sigma:g}, edge = {edge:.3g}: {escaped:.0%} escaped")
    plt.show()


if WIDGETS_AVAILABLE:
    display(widgets.interactive(
        explore,
        sigma=widgets.FloatSlider(value=2.0, min=0.0, max=4.0, step=0.5, description="sigma"),
        log10_edge=widgets.FloatSlider(value=-2.3, min=-3, max=0.3, step=0.1, description="log10(edge)"),
    ))
else:
    for sigma, le in [(2.0, -2.3), (2.0, -0.3), (0.0, -2.3)]:
        explore(sigma, le)
```

---
## Takeaways
#
1. With noise, NM compares values whose true difference can be far smaller than the noise;
   a small simplex then **stagnates wherever it starts** - on a peak or on a slope. That is
   stagnation, not convergence to a maximizer.
2. NM reports its luckiest evaluation: the final value is biased low. Re-evaluate the
   returned point before trusting it.
3. Size the initial simplex to the noise level, and use averaging/re-evaluation within the
   evaluation budget; for serious noisy problems use methods designed for noise.
