Companion notebook to the *Heuristics* section of the lecture (simulated annealing, evolution
strategies, genetic algorithms). Heuristics trade convergence guarantees for **global

local method gets trapped, and measure their cost and their run-to-run variability.

## 1. A multimodal test function
#
The Rastrigin function on the box $\Omega = [-5.12, 5.12]^2$,
#
$$
f(x) = 20 + \sum_{i=1}^{2}\big(x_i^2 - 10\cos(2\pi x_i)\big),
$$
#
has its global minimum $f = 0$ at the origin and a local minimum near every integer point.
A local method such as Nelder-Mead stops in the nearest basin.

```python
LOWER, UPPER = -5.12, 5.12


def rastrigin(x):
    x = np.asarray(x, dtype=float)
    return 20.0 + np.sum(x ** 2 - 10.0 * np.cos(2 * np.pi * x))


x_start = np.array([3.6, -2.7])
res_nm = minimize(rastrigin, x_start, method="Nelder-Mead")
print(f"Nelder-Mead from {x_start}: x = {np.round(res_nm.x, 3)}, f = {res_nm.fun:.3f}, evaluations = {res_nm.nfev}")

grid = np.linspace(LOWER, UPPER, 300)
GX, GY = np.meshgrid(grid, grid)
GZ = 20 + GX ** 2 - 10 * np.cos(2 * np.pi * GX) + GY ** 2 - 10 * np.cos(2 * np.pi * GY)
```

---
## 2. Simulated annealing
#
From the slides: choose a random neighbour $x' \in N(x) \cap \Omega$ and a random
$r \in [0, 1]$; move to $x'$ if $r < p(T_k, x, x') = e^{(f(x) - f(x'))/T_k}$; then cool the
temperature, $T_{k+1} = \alpha T_k$. Improving moves are always accepted ($p > 1$);
worsening moves are accepted often when $T$ is large and almost never when $T$ is small.
Our neighbourhood is a Gaussian step, and neighbours outside the box are redrawn.

```python
def simulated_annealing(func, x0, T0=100.0, alpha=0.998, step=0.25, budget=3000, seed=0):
    rng = np.random.default_rng(seed)
    x = np.array(x0, dtype=float)
    fx = func(x)
    best_x, best_f = x.copy(), fx
    T = T0
    trace = {"x": [x.copy()], "f": [fx], "best": [best_f], "T": [T], "accepted_worse": 0}
    for _ in range(budget - 1):
        while True:
            x_new = x + step * rng.normal(size=x.size)
            if np.all((x_new >= LOWER) & (x_new <= UPPER)):
                break
        f_new = func(x_new)
        if rng.random() < np.exp(min(0.0, (fx - f_new) / T)):
            trace["accepted_worse"] += f_new > fx
            x, fx = x_new, f_new
        if fx < best_f:
            best_x, best_f = x.copy(), fx
        T *= alpha
        for key, val in (("x", x.copy()), ("f", fx), ("best", best_f), ("T", T)):
            trace[key].append(val)
    return best_x, best_f, trace


x_sa, f_sa, tr_sa = simulated_annealing(rastrigin, x_start, seed=1)
print(f"SA: best x = {np.round(x_sa, 3)}, best f = {f_sa:.3f}, accepted worse moves = {tr_sa['accepted_worse']}")

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].contour(GX, GY, GZ, levels=20, cmap="Greys", linewidths=0.5)
path = np.array(tr_sa["x"])
sc = axes[0].scatter(path[:, 0], path[:, 1], c=np.arange(len(path)), cmap="plasma", s=4)
axes[0].plot(*x_start, "s", color="k", ms=8, label="start")
axes[0].plot(0, 0, "*", color=COLORS["optimum"], ms=16, mec="k", label="global min")
axes[0].set(xlabel="$x_1$", ylabel="$x_2$", aspect="equal", title="SA trajectory (colour = iteration)")
axes[0].legend(loc="upper left", fontsize=8)
plt.colorbar(sc, ax=axes[0], label="iteration")
axes[1].plot(tr_sa["f"], color=COLORS["sa"], lw=0.6, label="current $f(x)$")
axes[1].plot(tr_sa["best"], color="k", lw=2, label="best so far")
ax2 = axes[1].twinx()
ax2.semilogy(tr_sa["T"], color=COLORS["ga"], lw=1.5)
ax2.set_ylabel("temperature $T$", color=COLORS["ga"])
axes[1].set(xlabel="iteration", ylabel="$f$", title="Diversification (high T) then intensification")
axes[1].legend(loc="upper right", fontsize=8)
plt.tight_layout()
plt.show()
```

While $T$ is large, SA wanders across basins and accepts many worse points
(**diversification**); as $T \to 0$ it behaves like a greedy local search inside one basin
(**intensification**).
#
> **Predict:** what happens if we start with $T_0$ very small? Very large?

```python
for T0 in (0.01, 1.0, 100.0, 1000.0):
    finals = [simulated_annealing(rastrigin, x_start, T0=T0, seed=s)[1] for s in range(20)]
    print(f"step 0.25, T0 = {T0:7.2f}: median best f over 20 runs = {np.median(finals):6.3f}, "
          f"runs with f < 1.1: {sum(v < 1.1 for v in finals)}/20")
finals = [simulated_annealing(rastrigin, x_start, T0=0.01, step=0.6, seed=s)[1] for s in range(20)]
print(f"step 0.60, T0 =    0.01: median best f over 20 runs = {np.median(finals):6.3f}, "
      f"runs with f < 1.1: {sum(v < 1.1 for v in finals)}/20")
```

With a local neighbourhood (step $0.25$), a tiny $T_0$ turns SA into a greedy local search
that never leaves the starting basin ($f \approx 15$); a large $T_0$ lets it cross the
barriers between basins before cooling down. The last line shows the other diversification
knob: with a **large neighbourhood** (step $0.6$), even greedy acceptance jumps between
basins. Temperature, cooling rate and step size interact, and their best values depend on
the scale of $f$ - this is the "intense parameter tuning" the slides warn about.

## 3. A genetic algorithm with generational replacement
#
The slides' GA: start from a population $X_0$ of size $\lambda$; at every generation, create
$\lambda$ offspring, each from 2 individuals of $X_{k-1}$ (here: **tournament selection**,
**blend crossover** and **Gaussian mutation**), and replace the whole population by the
offspring. Offspring outside the box are **repaired** by clipping (one of the four
infeasibility options in the slides).

```python
def genetic_algorithm(func, pop_size=30, n_gen=100, p_mut=0.2, sigma_mut=0.5, seed=0, elitism=False):
    rng = np.random.default_rng(seed)
    pop = rng.uniform(LOWER, UPPER, size=(pop_size, 2))
    fit = np.array([func(p) for p in pop])
    best_x, best_f = pop[np.argmin(fit)].copy(), fit.min()
    history = {"pop": [pop.copy()], "best": [best_f], "pop_best": [fit.min()], "n_eval": [pop_size]}

    def tournament():
        i, j = rng.integers(pop_size, size=2)
        return pop[i] if fit[i] < fit[j] else pop[j]

    for _ in range(n_gen):
        children = []
        for _ in range(pop_size):
            p1, p2 = tournament(), tournament()
            w = rng.uniform(-0.25, 1.25, size=2)                 # blend crossover
            child = w * p1 + (1 - w) * p2
            mutate = rng.random(2) < p_mut
            child = child + mutate * rng.normal(0.0, sigma_mut, size=2)
            children.append(np.clip(child, LOWER, UPPER))          # repair
        children = np.array(children)
        child_fit = np.array([func(c) for c in children])
        if elitism:                                                # keep the best parent
            worst = np.argmax(child_fit)
            children[worst], child_fit[worst] = pop[np.argmin(fit)], fit.min()
        pop, fit = children, child_fit                             # generational replacement
        if fit.min() < best_f:
            best_x, best_f = pop[np.argmin(fit)].copy(), fit.min()
        history["pop"].append(pop.copy())
        history["best"].append(best_f)
        history["pop_best"].append(fit.min())
        history["n_eval"].append(history["n_eval"][-1] + pop_size)
    return best_x, best_f, history


x_ga, f_ga, hist_ga = genetic_algorithm(rastrigin, seed=1)
print(f"GA: best x = {np.round(x_ga, 3)}, best f = {f_ga:.3f}, evaluations = {hist_ga['n_eval'][-1]}")

fig, axes = plt.subplots(1, 4, figsize=(15, 3.9))
for ax, gen in zip(axes, (0, 5, 20, 100)):
    ax.contour(GX, GY, GZ, levels=15, cmap="Greys", linewidths=0.4)
    ax.scatter(*hist_ga["pop"][gen].T, s=12, color=COLORS["pop"])
    ax.plot(0, 0, "*", color=COLORS["optimum"], ms=12, mec="k")
    ax.set(xlim=(LOWER, UPPER), ylim=(LOWER, UPPER), aspect="equal", title=f"generation {gen}")
plt.tight_layout()
plt.show()
```

# The population starts spread over the whole box (diversification) and gradually
# concentrates in the best basins (intensification), driven by selection pressure. Mutation
# keeps injecting diversity.
#
# With pure generational replacement the best individual can be **lost** from one generation
# to the next: the best-so-far value (which we record separately) is not the same as the best
# value in the current population.

```python
_, _, hist_plain = genetic_algorithm(rastrigin, seed=3, elitism=False)
_, _, hist_elite = genetic_algorithm(rastrigin, seed=3, elitism=True)
fig, ax = plt.subplots(figsize=(7.5, 4.2))
ax.semilogy(hist_plain["n_eval"], np.maximum(hist_plain["pop_best"], 1e-8), color=COLORS["ga"], lw=1,
            label="best in current population (generational replacement)")
ax.semilogy(hist_plain["n_eval"], np.maximum(hist_plain["best"], 1e-8), color="k", lw=2, label="best so far")
ax.semilogy(hist_elite["n_eval"], np.maximum(hist_elite["pop_best"], 1e-8), color=COLORS["sa"], lw=1.5,
            ls="--", label="best in population, with elitism")
ax.set(xlabel="function evaluations", ylabel="$f$", title="GA: the population can forget its best point")
ax.legend(fontsize=8)
plt.show()
```

## 4. Many runs: cost and lack of convergence guarantees
#
The slides warn that heuristics "spend a lot of evaluations and require intense parameter
tuning" and have no convergence properties. We give NM (with random restarts), SA and GA the
same budget of 3000 evaluations and repeat each experiment 30 times.

```python
def nm_restarts(seed, budget=3000):
    rng = np.random.default_rng(seed)
    best, used = np.inf, 0
    while used < budget:
        res = minimize(rastrigin, rng.uniform(LOWER, UPPER, 2), method="Nelder-Mead",
                       options={"maxfev": budget - used})
        used += res.nfev
        best = min(best, res.fun)
    return best


results = {
    "NM, one start": [minimize(rastrigin, np.random.default_rng(s).uniform(LOWER, UPPER, 2),
                               method="Nelder-Mead").fun for s in range(30)],
    "NM with restarts": [nm_restarts(s) for s in range(30)],
    "simulated annealing": [simulated_annealing(rastrigin, np.random.default_rng(s).uniform(LOWER, UPPER, 2),
                                                seed=s)[1] for s in range(30)],
    "genetic algorithm": [genetic_algorithm(rastrigin, seed=s, n_gen=99)[1] for s in range(30)],
}
fig, ax = plt.subplots(figsize=(8, 4.2))
ax.boxplot([np.maximum(v, 1e-6) for v in results.values()], showfliers=True)
ax.set_xticks(range(1, len(results) + 1), labels=list(results.keys()))
ax.set(yscale="log", ylabel="best $f$ found (30 runs)", title="Same budget (3000 evaluations), 30 seeds")
plt.show()
for name, vals in results.items():
    print(f"{name:20s}: median {np.median(vals):.3f}, runs with f < 0.1: {sum(v < 0.1 for v in vals)}/30")
```

A single local search ends in a random basin. The heuristics and the multistart strategy
all find much better points, but the outcome **varies from run to run** - nothing guarantees
the global minimum, or even a stationary point, for a given budget. On this smooth function
a multistart local method is a strong competitor; heuristics pay off on rugged, nonsmooth,
discrete or mixed problems where local methods struggle.

## 5. Explore: the SA temperature schedule

```python
def explore_sa(log10_T0=2.0, alpha=0.998, step=0.25):
    xb, fb, tr = simulated_annealing(rastrigin, x_start, T0=10 ** log10_T0, alpha=alpha, step=step, seed=2)
    path = np.array(tr["x"])
    fig, ax = plt.subplots(figsize=(5.5, 5))
    ax.contour(GX, GY, GZ, levels=20, cmap="Greys", linewidths=0.5)
    ax.scatter(path[:, 0], path[:, 1], c=np.arange(len(path)), cmap="plasma", s=4)
    ax.plot(*xb, "o", color=COLORS["sa"], ms=9, mec="k")
    ax.set(aspect="equal", title=f"T0 = {10 ** log10_T0:.3g}, alpha = {alpha}: best f = {fb:.3f}")
    plt.show()


if WIDGETS_AVAILABLE:
    display(widgets.interactive(
        explore_sa,
        log10_T0=widgets.FloatSlider(value=2.0, min=-2, max=3, step=0.5, description="log10(T0)"),
        alpha=widgets.FloatSlider(value=0.998, min=0.9, max=0.999, step=0.001, readout_format=".3f"),
        step=widgets.FloatSlider(value=0.25, min=0.05, max=2.0, step=0.05),
    ))
else:
    for T0 in (-2.0, 2.0, 3.0):
        explore_sa(T0)
```

## Takeaways
#
1. SA accepts worse moves with probability $e^{(f(x) - f(x'))/T}$; a high temperature
   diversifies, cooling ($T \leftarrow \alpha T$) intensifies.
2. A GA evolves a population by selection, crossover and mutation; with generational
   replacement the best point can be lost unless it is stored or kept (elitism).
3. Both methods explore globally, use many evaluations and depend strongly on tuning
   ($T_0$, $\alpha$, step, population size, mutation rate).
4. Results vary between runs and come with no convergence guarantee: compare heuristics
   over many seeds and against simple baselines such as a multistart local method.
