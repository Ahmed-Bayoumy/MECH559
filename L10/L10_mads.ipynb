# Mesh Adaptive Direct Search (MADS)
### MECH 559 - Systems Optimization - L10
### Course Instructor: Dr. Ahmed BayoumyAuthor of the notebook: Dr. Ahmed Bayoumy
Companion notebook to the *Mesh adaptive direct search* section of the lecture. MADS keeps
the Search/Poll structure of GPS but separates two sizes: the **mesh size** $\delta^k$ (how
fine the grid of admissible points is) and the **frame size** $\Delta^k$ (how far the poll
points are from the incumbent), with $\delta^k \le \Delta^k$ and $\delta^k$ shrinking faster.
This freedom lets MADS choose new poll directions at every iteration, so that they become
**asymptotically dense** in the unit sphere. We build the poll directions with Householder
matrices (as in OrthoMADS-type implementations), check their density, and show that MADS
solves the nonsmooth problems on which coordinate search and GPS stall.

## Learning objectives

By the end, students should be able to:

1. Construct the MADS mesh $M^k$ and frame $F^k$ from $\delta^k = \min\{\Delta^k, (\Delta^k)^2\}$
   and $\Delta^k$, and explain why $\delta^k/\Delta^k \to 0$ gives more poll directions.
2. Build a positive spanning set of poll directions from the Householder matrix
   $H = I - 2vv^\top$ of a random unit vector $v$.
3. Show numerically that the normalized poll directions become dense on the unit circle.
4. Use MADS with the extreme barrier on a constrained problem, and know where to find
   production implementations (NOMAD, OMADS).

## 1. Mesh and frame
#
With frame size $\Delta^k$ and mesh size $\delta^k = \min\{\Delta^k, (\Delta^k)^2\}$:
#
- the **mesh** $M^k = \{x^k + \delta^k D y : y \in \mathbb{N}^p\}$ is the grid of points MADS may evaluate;
- the **frame** $F^k = \{x \in M^k : \|x - x^k\|_\infty \le \Delta^k b\}$ is where the poll points live.
#
GPS has $\delta^k = \Delta^k$: the frame contains only the immediate neighbours. In MADS, when
$\Delta^k$ is halved, $\delta^k$ is divided by four, so the frame contains more and more mesh
points - i.e. more and more possible poll directions.

```python
def poll_directions(Delta, rng):
    """MADS poll directions from a Householder matrix (2n directions, integer on the mesh)."""
    n = 2
    delta = min(Delta, Delta ** 2)
    v = rng.normal(size=n)
    v /= np.linalg.norm(v)
    H = np.eye(n) - 2.0 * np.outer(v, v)
    B = np.round((Delta / delta) * H / np.max(np.abs(H), axis=0))
    return np.vstack([B.T, -B.T]), delta


rng = np.random.default_rng(3)
fig, axes = plt.subplots(1, 3, figsize=(14, 4.6))
for ax, Delta in zip(axes, (1.0, 0.5, 0.25)):
    D, delta = poll_directions(Delta, rng)
    ax.grid(False)
    for v in np.arange(-1.0, 1.0 + 1e-9, delta):
        ax.axvline(v, color=COLORS["mesh"], lw=0.6)
        ax.axhline(v, color=COLORS["mesh"], lw=0.6)
    ax.add_patch(plt.Rectangle((-Delta, -Delta), 2 * Delta, 2 * Delta, fill=False, lw=2.5, color=COLORS["frame"]))
    for d in D:
        p = delta * d
        ax.plot([0, p[0]], [0, p[1]], color=COLORS["mads"], lw=1.5)
        ax.plot(*p, "o", color=COLORS["mads"], ms=6)
    ax.plot(0, 0, "s", color="k", ms=7)
    ax.set(xlim=(-1.05, 1.05), ylim=(-1.05, 1.05), aspect="equal",
           title=f"$\\Delta^k$ = {Delta:g} (frame), $\\delta^k$ = {delta:g} (mesh)")
plt.tight_layout()
plt.show()
```

The thick square is the frame; the grey grid is the mesh; red points are the poll points.
As $\Delta^k$ decreases, the mesh becomes much finer than the frame and the poll directions
can point almost anywhere.

## 2. Householder poll directions are dense
#
For a random unit vector $v$, the Householder matrix $H = I - 2vv^\top$ is orthogonal, so its
columns $h_j$ and their negatives form a maximal positive basis. Scaling each column so that
its largest component is $\Delta^k/\delta^k$ and rounding puts the poll points on the mesh,
at a distance $\approx \Delta^k$ from $x^k$.
#
> **Predict:** after 200 iterations with decreasing frame size, how will the set of
> normalized directions used by MADS compare with GPS's four directions?

```python
rng = np.random.default_rng(0)
used = []
Delta = 1.0
for k in range(200):
    D, _ = poll_directions(Delta, rng)
    used.extend(D / np.linalg.norm(D, axis=1, keepdims=True))
    Delta = max(Delta / 2, 1e-6) if k % 3 else Delta     # mimic a sequence of failures/successes
used = np.array(used)
angles = np.mod(np.arctan2(used[:, 1], used[:, 0]), 2 * np.pi)
largest_gap = np.max(np.diff(np.sort(np.concatenate([angles, [angles.min() + 2 * np.pi]]))))

fig = plt.figure(figsize=(11, 4.6))
ax1 = fig.add_subplot(1, 2, 1)
ax1.scatter(used[:, 0], used[:, 1], s=6, color=COLORS["mads"], alpha=0.5, label="MADS directions")
for d in ([1, 0], [0, 1], [-1, 0], [0, -1]):
    ax1.annotate("", xy=d, xytext=(0, 0), arrowprops=dict(arrowstyle="->", color=COLORS["gps"], lw=2.5))
ax1.set(aspect="equal", xlim=(-1.2, 1.2), ylim=(-1.2, 1.2), title="Normalized poll directions (blue: GPS)")
ax2 = fig.add_subplot(1, 2, 2, projection="polar")
ax2.hist(angles, bins=72, color=COLORS["mads"], alpha=0.8)
ax2.set_title(f"Direction angles: largest gap = {np.degrees(largest_gap):.1f} deg", pad=15)
plt.tight_layout()
plt.show()
```

GPS uses the same 4 directions forever (largest angular gap $90^\circ$); MADS's directions
keep filling the circle, so **every** cone of descent directions, however narrow and however
oriented, is eventually hit. This is what upgrades the convergence result from "no descent
along the directions in $D$" (GPS) to **Clarke stationarity** (MADS, for $f$ Lipschitz near
the limit).

## 3. MADS on the kink functions that stop CS and GPS
#
A basic MADS loop: poll opportunistically along the Householder directions; on success
$\Delta^{k+1} = 2\Delta^k$, on failure $\Delta^{k+1} = \Delta^k / 2$.

```python
def mads(func, x0, Delta0=1.0, budget=2000, tol=1e-9, seed=0):
    rng = np.random.default_rng(seed)
    x, fx, Delta = np.array(x0, dtype=float), func(x0), Delta0
    n_eval, hist = 1, [(1, np.array(x0, dtype=float), func(x0), Delta0)]
    while n_eval < budget and Delta > tol:
        D, delta = poll_directions(Delta, rng)
        success = False
        for d in D:
            t = x + delta * d
            ft = func(t)
            n_eval += 1
            if ft < fx:
                x, fx, success = t, ft, True
                break
        Delta = 2 * Delta if success else Delta / 2
        hist.append((n_eval, x.copy(), fx, Delta))
    return x, fx, hist


def gps(func, x0, D, delta0=1.0, budget=2000, tol=1e-9):
    x, fx, delta, n_eval = np.array(x0, dtype=float), func(x0), delta0, 1
    while n_eval < budget and delta > tol:
        for d in D:
            t = x + delta * d
            ft = func(t)
            n_eval += 1
            if ft < fx:
                x, fx = t, ft
                break
        else:
            delta /= 2
    return x, fx


def g(x):
    return abs(x[0] - x[1]) + 0.5 * abs(x[0] + x[1])


def g2(x):
    return abs(x[0] + x[1]) + 0.5 * abs(x[0] - x[1])


D_coord = np.array([[1, 0], [0, 1], [-1, 0], [0, -1]], dtype=float)
D_min = np.array([[1, 0], [0, 1], [-1, -1]], dtype=float)
rows = {}
for name, func, start in [("g from (1, 1)", g, (1.0, 1.0)), ("g2 from (1, -1)", g2, (1.0, -1.0))]:
    mads_final = [mads(func, start, seed=s)[1] for s in range(30)]
    rows[name] = {"CS (coordinate D)": gps(func, start, D_coord)[1],
                  "GPS (minimal D)": gps(func, start, D_min)[1],
                  "MADS: worst of 30 seeds": max(mads_final)}
pd.DataFrame(rows).T
```

# Both fixed direction sets stop at a non-minimizer on at least one of the two problems, while
# every MADS run reaches the minimizer (the final values are at round-off level).

```python
x_m, f_m, hist_m = mads(g2, (1.0, -1.0), seed=1)
path = np.array([h[1] for h in hist_m])
gx = np.linspace(-1.2, 1.2, 301)
GX, GY = np.meshgrid(gx, gx)
fig, axes = plt.subplots(1, 2, figsize=(12, 4.8))
axes[0].contour(GX, GY, np.abs(GX + GY) + 0.5 * np.abs(GX - GY), levels=20, cmap="Greys", linewidths=0.6)
axes[0].plot(path[:, 0], path[:, 1], "o-", ms=3, color=COLORS["mads"], label="MADS incumbents")
axes[0].plot(1, -1, "s", color=COLORS["gps"], ms=9, label="CS and GPS stop here")
axes[0].plot(0, 0, "*", color=COLORS["optimum"], ms=16, mec="k", label="minimizer")
axes[0].set(aspect="equal", xlabel="$x_1$", ylabel="$x_2$", title="$g_2$: MADS escapes the kink")
axes[0].legend(fontsize=8)
axes[1].semilogy([h[0] for h in hist_m], [max(h[2], 1e-16) for h in hist_m], color=COLORS["mads"], lw=2, label="$f(x^k)$")
axes[1].semilogy([h[0] for h in hist_m], [h[3] for h in hist_m], color=COLORS["frame"], lw=1.5, label="frame size $\\Delta^k$")
axes[1].set(xlabel="function evaluations", title="Progress of MADS on $g_2$")
axes[1].legend()
plt.tight_layout()
plt.show()
```

## 4. Constraints with the extreme barrier
#
MADS handles constraints by optimizing $f_\Omega$ ($+\infty$ outside $\Omega$), which also
covers hidden constraints (crashed simulations). Problem:
#
$$
\min_x (x_1 - 3)^2 + (x_2 - 2)^2 \quad \text{s.t.} \quad x_1^2 + x_2^2 \le 4,
$$
#
whose solution is the projection of $(3, 2)$ on the disk: $x^\star = 2\,(3, 2)/\sqrt{13}$.
#
> **Predict:** the optimum is on the boundary, where the barrier jumps to $+\infty$. Can a
> method that never sees an infeasible value approach it?

```python
def f_constrained(x):
    if x[0] ** 2 + x[1] ** 2 > 4.0:
        return np.inf                                     # extreme barrier
    return (x[0] - 3.0) ** 2 + (x[1] - 2.0) ** 2


x_true = 2 * np.array([3.0, 2.0]) / np.sqrt(13)
finals = [mads(f_constrained, (0.0, 0.0), budget=3000, tol=1e-10, seed=s) for s in range(20)]
errs = [np.linalg.norm(r[0] - x_true) for r in finals]
print(f"x* = {np.round(x_true, 6)}, f* = {f_constrained(x_true):.6f}")
print(f"MADS over 20 seeds: median ||x - x*|| = {np.median(errs):.1e}, worst = {max(errs):.1e}")
xg, fg = gps(f_constrained, (0.0, 0.0), D_coord, budget=3000, tol=1e-10)
print(f"GPS (coordinate D): x = {np.round(xg, 6)}, ||x - x*|| = {np.linalg.norm(xg - x_true):.1e}")

x_b, _, hist_b = finals[0]
path_b = np.array([h[1] for h in hist_b])
fig, ax = plt.subplots(figsize=(5.5, 5))
th = np.linspace(0, 2 * np.pi, 200)
ax.fill(2 * np.cos(th), 2 * np.sin(th), color=COLORS["feasible"], alpha=0.5, label="$\\Omega$")
xs = np.linspace(-2.5, 3.5, 200)
XX, YY = np.meshgrid(xs, xs)
ax.contour(XX, YY, (XX - 3) ** 2 + (YY - 2) ** 2, levels=15, cmap="Greys", linewidths=0.6)
ax.plot(path_b[:, 0], path_b[:, 1], "o-", ms=3, color=COLORS["mads"], label="MADS incumbents")
ax.plot(*x_true, "*", color=COLORS["optimum"], ms=16, mec="k", label="$x^\\star$")
ax.plot(*xg, "s", color=COLORS["gps"], ms=9, label="GPS stops here")
ax.set(aspect="equal", xlim=(-2.5, 3.5), ylim=(-2.5, 3.5), title="Extreme barrier: all incumbents feasible")
ax.legend(fontsize=8, loc="lower right")
plt.show()
```

All incumbents stay feasible. **GPS stalls at $(2, 0)$**, far from $x^\star$: there, the
move $+e_2$ is infeasible (the boundary curves away) and the other coordinate moves go uphill,
although feasible descent directions exist (slightly inward of the tangent). This is the
kink problem again: the extreme barrier creates a "kink" at the boundary.
#
MADS gets much closer, but **not** to round-off accuracy: near $x^\star$ the cone of
feasible descent directions becomes very narrow, successful polls become rare, and the frame
size reaches the tolerance first (each run stops after only a few hundred evaluations, so a
larger budget does not help). The MADS convergence result is asymptotic ($\Delta^k \to 0$
with infinitely many polls); practical implementations accelerate it with Search steps
(e.g. quadratic models) and, when infeasible points carry information (relaxable
constraints), with the **progressive barrier** of NOMAD, which exploits the constraint
violation instead of rejecting it.

## 5. Production implementations
#
The loop above is a teaching version. Real implementations add a Search step, an
opportunistic and ordered poll, caching, constraint handling with the progressive barrier,
variable types and parallelism:
#
- **NOMAD** (C++, with Python interface `PyNomadBBO`): the state of the art,
  https://nomad-4-user-guide.readthedocs.io/
- **OMADS** (Python): https://ahmed-bayoumy.github.io/OMADS/
- **NOMADm** (MATLAB), linked from the lecture slides.

```python
def explore_mads(log10_Delta=0.0, seed=0):
    Delta = 10 ** log10_Delta
    rng_ = np.random.default_rng(seed)
    D, delta = poll_directions(Delta, rng_)
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.grid(False)
    if Delta / delta <= 64:
        for v in np.arange(-Delta * 1.2, Delta * 1.2 + 1e-12, delta):
            ax.axvline(v, color=COLORS["mesh"], lw=0.5)
            ax.axhline(v, color=COLORS["mesh"], lw=0.5)
    ax.add_patch(plt.Rectangle((-Delta, -Delta), 2 * Delta, 2 * Delta, fill=False, lw=2.5, color=COLORS["frame"]))
    for d in D:
        p = delta * d
        ax.plot([0, p[0]], [0, p[1]], color=COLORS["mads"], lw=1.5)
        ax.plot(*p, "o", color=COLORS["mads"], ms=6)
    ax.set(xlim=(-1.25 * Delta, 1.25 * Delta), ylim=(-1.25 * Delta, 1.25 * Delta), aspect="equal",
           title=f"Delta = {Delta:.3g}, delta = {delta:.3g}, ratio = {Delta / delta:.0f}")
    plt.show()


if WIDGETS_AVAILABLE:
    display(widgets.interactive(
        explore_mads,
        log10_Delta=widgets.FloatSlider(value=0.0, min=-3, max=0, step=0.30103, description="log10(Delta)"),
        seed=widgets.IntSlider(value=0, min=0, max=20, description="seed (v)"),
    ))
else:
    for value in (0.0, -0.60206, -1.20412):
        explore_mads(value)
```

## Takeaways
#
1. MADS separates the mesh size $\delta^k$ from the frame size $\Delta^k$; since
   $\delta^k/\Delta^k \to 0$, the frame offers more and more candidate poll directions.
2. Householder matrices of random unit vectors give a maximal positive basis at every
   iteration; over the run, the normalized directions become dense on the unit sphere.
3. Dense directions eventually hit every cone of descent directions, which is why MADS
   solves nonsmooth problems on which any fixed direction set (CS, GPS) can stall, and why its
   limit points are Clarke stationary when $f$ is Lipschitz near them.
4. Constraints, including hidden ones, are handled with the extreme barrier (or the
   progressive barrier in NOMAD); use NOMAD or OMADS for real problems.
