Companion notebook to the *MADS implementations* slides. The previous L10 notebooks built
coordinate search, GPS and MADS from scratch to understand **how** they work. In practice you
should use a mature implementation. **OMADS** is a Python implementation of the Mesh Adaptive
Direct Search algorithm (Search + Poll, orthogonal or granular mesh, progressive and extreme
barriers). This notebook is a set of tutorials, each solving a different kind of
single-objective optimization (SOO) problem:

| Tutorial | Problem type | OMADS feature |
|---|---|---|
| 1 | smooth, unconstrained (Rosenbrock) | `poll` vs `mads` (Search + Poll) |
| 2 | nonsmooth (a rotated kink) | rich (Householder) poll directions |
| 3 | inequality constraints | progressive (PB) vs extreme (EB) barrier |
| 4 | hidden constraints (crashing simulations) | returning $+\infty$ / NaN |
| 5 | 10 variables, 6 constraints (slides' geometric programming problem) | a realistic constrained run |
| 6 | 10-dimensional Rosenbrock | scaling up |

## 0. Anatomy of an OMADS problem
#
An OMADS run is described by a dictionary with four parts:
#
- `evaluator`: the **blackbox**, a Python callable taking the list of variables `x` and
  returning `[f, [c_1, ..., c_m]]` - the objective and the constraint values, where
  **$c_j(x) \le 0$ means feasible**. An unconstrained problem returns `[f, [0]]`.
- `param`: the problem - starting point `baseline`, bounds `lb`/`ub`, `var_names`,
  `scaling` (poll steps are measured in units of $(ub - lb)/\text{scaling}$),
  `constraints_type` (`"PB"` or `"EB"` per constraint), `name` and `post_dir`.
- `options`: the algorithm - `seed`, evaluation `budget`, frame-size `tol`,
  `rich_direction` (random Householder poll directions, as in the MADS notebook),
  `opportunistic` (stop a poll as soon as it succeeds), and cache settings.
- `search` (for the `search` and `mads` solvers): the Search step, e.g. `"sampling"` with
  `"ACTIVE"` sampling of `ns` points.
#
OMADS prints a progress bar at every evaluation. The helper below silences it, records every
blackbox call, and returns the OMADS summary together with an evaluation history.

```python
BASE_OPTIONS = {"seed": 0, "budget": 1000, "tol": 1e-12, "display": False, "save_results": False,
                "check_cache": True, "store_cache": True, "rich_direction": True,
                "opportunistic": False, "is_verbose": False}
BASE_SEARCH = {"type": "sampling", "s_method": "ACTIVE", "ns": 10, "visualize": False}
SOLVERS = {"poll": poll, "search": search, "mads": mads}


def solve(blackbox, param, solver="mads", options=None, search_options=None):
    """Run an OMADS solver quietly; return (summary dict, history DataFrame, wall time)."""
    calls = []

    def logged_blackbox(x):                    # OMADS requires a single-argument callable
        out = blackbox(x)
        calls.append((list(x), out))
        return out

    data = {"evaluator": {"blackbox": logged_blackbox},
            "param": {"post_dir": POST_DIR, **param},
            "options": {**BASE_OPTIONS, **(options or {})},
            "search": {**BASE_SEARCH, **(search_options or {})}}
    tic = time.perf_counter()
    with contextlib.redirect_stdout(io.StringIO()):
        summary = SOLVERS[solver].main(data)[0]
    wall = time.perf_counter() - tic

    rows = []
    for x, (f, c) in calls:
        f = float(np.ravel(f)[0])
        h = float(np.sum(np.maximum(np.asarray(c, dtype=float), 0.0) ** 2))   # constraint violation
        rows.append({"x": x, "f": f, "h": h, "feasible": h == 0.0 and np.isfinite(f)})
    hist = pd.DataFrame(rows)
    hist["best feasible f"] = hist["f"].where(hist["feasible"]).cummin()
    return summary, hist, wall


def report(name, summary, hist, wall):
    return {"run": name, "fmin": summary["fmin"][0], "hmin": summary["hmin"],
            "xmin": np.round(summary["xmin"], 5).tolist(), "blackbox calls": len(hist), "time [s]": round(wall, 1)}
```

## Tutorial 1. Smooth unconstrained problem: Rosenbrock
#
$$
\min_{x \in [-5, 10]^2} f(x) = 100(x_2 - x_1^2)^2 + (1 - x_1)^2, \qquad x^\star = (1, 1).
$$
#
We compare the three solvers: `poll` (MADS Poll steps only), `search` (Search steps only)
and `mads` (Search, then Poll when the Search fails), with the same budget of 1000 calls.
#
> **Predict:** which solver reaches $f < 10^{-6}$ first?

```python
def rosenbrock(x):
    x = np.asarray(x, dtype=float)
    return [float(np.sum(100.0 * (x[1:] - x[:-1] ** 2) ** 2 + (1 - x[:-1]) ** 2)), [0.0]]


param_rb = {"name": "RB", "baseline": [-2.0, -2.0], "lb": [-5, -5], "ub": [10, 10],
            "var_names": ["x1", "x2"], "scaling": 10.0}

runs_rb = {s: solve(rosenbrock, param_rb, solver=s) for s in ("poll", "search", "mads")}
runs_rb["search (opportunistic)"] = solve(rosenbrock, param_rb, solver="search", options={"opportunistic": True})
pd.DataFrame([report(k, *v) for k, v in runs_rb.items()])
```

```python
fig, ax = plt.subplots(figsize=(7.5, 4.5))
for name, (summary, hist, _) in runs_rb.items():
    color = COLORS.get(name.split()[0])
    ax.semilogy(np.arange(1, len(hist) + 1), np.maximum(hist["best feasible f"], 1e-16),
                color=color, lw=2, ls="--" if "opportunistic" in name else "-", label=name)
ax.set(xlabel="blackbox evaluations", ylabel="best $f$ so far", title="Rosenbrock: OMADS solvers")
ax.legend()
plt.show()
```

The Poll step alone makes steady but slow progress along the curved valley. Adding the
Search step (`mads`), which proposes promising points from samples of the evaluated set,
reaches $f \approx 10^{-12}$ within about 500 evaluations. The `search` solver is sensitive to
its settings: with `opportunistic=True` it is excellent here, without it it stalls. **Use
`mads` as the default solver.**

## Tutorial 2. A nonsmooth problem: rich poll directions
#
The rotated kink from the coordinate search and GPS notebooks,
#
$$
g(x) = |x_1 + x_2| + \tfrac12 |x_1 - x_2|, \qquad x^0 = (1, -1) \text{ (on the kink)}, \quad x^\star = (0, 0),
$$
#
stops every fixed direction set we tried. OMADS polls along $2n$ orthogonal directions; with
`rich_direction=False` they are the coordinate directions, with `rich_direction=True` they
come from a random Householder matrix at every iteration (asymptotically dense).

```python
def kink(x):
    return [abs(x[0] + x[1]) + 0.5 * abs(x[0] - x[1]), [0.0]]


param_kink = {"name": "kink", "baseline": [1.0, -1.0], "lb": [-3, -3], "ub": [3, 3],
              "var_names": ["x1", "x2"], "scaling": 10.0}
runs_kink = {f"{s}, rich_direction={rd}": solve(kink, param_kink, solver=s, options={"rich_direction": rd, "budget": 500})
             for s in ("poll", "mads") for rd in (False, True)}
pd.DataFrame([report(k, *v) for k, v in runs_kink.items()])
```

With coordinate directions the Poll alone never leaves the kink ($f = 1$, exactly the GPS
failure from the lecture), while rich directions reach the minimizer. The Search step of
`mads` also rescues the coordinate-direction run, because its sample points are not
restricted to the poll directions. For nonsmooth blackboxes keep `rich_direction=True`.

## Tutorial 3. Inequality constraints: progressive vs extreme barrier
#
$$
\min_x (x_1 - 3)^2 + (x_2 - 2)^2 \quad \text{s.t.} \quad c(x) = x_1^2 + x_2^2 - 4 \le 0,
\qquad x^\star = \tfrac{2}{\sqrt{13}}(3, 2).
$$
#
- **EB** (extreme barrier): infeasible points are simply rejected ($f_\Omega = +\infty$).
- **PB** (progressive barrier): the constraint violation $h(x) = \sum_j \max(c_j(x), 0)^2$ is
  used; infeasible points with small $h$ can serve as poll centres, and the threshold $h_{max}$
  is progressively decreased.
#
We start each barrier from a feasible point $(0, 0)$ and from an infeasible point $(3, 3)$.

```python
def disk_problem(x):
    return [(x[0] - 3) ** 2 + (x[1] - 2) ** 2, [x[0] ** 2 + x[1] ** 2 - 4]]


x_star_disk = 2 * np.array([3.0, 2.0]) / np.sqrt(13)
runs_disk = {}
for barrier in ("EB", "PB"):
    for x0 in ([0.0, 0.0], [3.0, 3.0]):
        param = {"name": "disk", "baseline": x0, "lb": [-5, -5], "ub": [5, 5], "var_names": ["x1", "x2"],
                 "scaling": 10.0, "constraints_type": [barrier]}
        runs_disk[f"{barrier}, x0={x0}"] = solve(disk_problem, param, options={"budget": 600})

table = pd.DataFrame([report(k, *v) for k, v in runs_disk.items()])
table["||xmin - x*||"] = [np.linalg.norm(np.array(v[0]["xmin"]) - x_star_disk) for v in runs_disk.values()]
table["infeasible calls"] = [int((~v[1]["feasible"]).sum()) for v in runs_disk.values()]
table
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
th = np.linspace(0, 2 * np.pi, 200)
for ax, key in zip(axes, ("EB, x0=[3.0, 3.0]", "PB, x0=[3.0, 3.0]")):
    _, hist, _ = runs_disk[key]
    X = np.array(hist["x"].tolist())
    ax.plot(2 * np.cos(th), 2 * np.sin(th), color="k", lw=1)
    ax.scatter(X[~hist["feasible"], 0], X[~hist["feasible"], 1], s=8, color=COLORS["infeasible"], label="infeasible")
    ax.scatter(X[hist["feasible"], 0], X[hist["feasible"], 1], s=8, color=COLORS["feasible"], label="feasible")
    ax.plot(*x_star_disk, "*", color=COLORS["optimum"], ms=16, mec="w", label="$x^\\star$")
    ax.plot(3, 3, "s", color="k", ms=7, label="start")
    ax.set(aspect="equal", xlim=(-3, 4), ylim=(-3, 4), title=f"{key}: evaluated points")
    ax.legend(fontsize=8, loc="lower left")
plt.tight_layout()
plt.show()
```

Both barriers find $x^\star$ to about $10^{-5}$ - even from the infeasible start - and the
progressive barrier is slightly more accurate. A large share of the evaluations is spent just
outside the boundary: that is where the optimum is. In the hand-made MADS of the previous
notebook the same problem stopped about $0.08$ away from $x^\star$; OMADS's Search step and
mesh management make the difference.

## Tutorial 4. Hidden constraints: when the simulation crashes
#
Suppose the constraint above is **hidden**: the simulation simply fails outside the disk.
OMADS treats a returned `inf` or `NaN` objective as a failed evaluation. An exception raised
inside the blackbox, however, stops the whole run - so wrap your simulation call.

```python
def run_simulation(x):
    if x[0] ** 2 + x[1] ** 2 > 4:
        raise RuntimeError("mesh generation failed")
    return (x[0] - 3) ** 2 + (x[1] - 2) ** 2


def safe_blackbox(x):
    try:
        return [run_simulation(x), [0.0]]
    except RuntimeError:
        return [np.inf, [0.0]]             # failed evaluation


param_hidden = {"name": "hidden", "baseline": [0.0, 0.0], "lb": [-5, -5], "ub": [5, 5],
                "var_names": ["x1", "x2"], "scaling": 10.0}
summary_h, hist_h, wall_h = solve(safe_blackbox, param_hidden, options={"budget": 600})
summary_k, hist_k, wall_k = runs_disk["PB, x0=[0.0, 0.0]"]
pd.DataFrame([
    {**report("hidden constraint (inf on crash)", summary_h, hist_h, wall_h),
     "||xmin - x*||": np.linalg.norm(np.array(summary_h["xmin"]) - x_star_disk)},
    {**report("same constraint returned as c(x) <= 0 (PB)", summary_k, hist_k, wall_k),
     "||xmin - x*||": np.linalg.norm(np.array(summary_k["xmin"]) - x_star_disk)},
])
```

OMADS copes with crashes (and stops earlier, once the frame size reaches the tolerance), but
the solution is less accurate: a crash only says "not here", while a returned constraint
value says "how far". **Whenever the simulation can measure a constraint, return it as
$c_j(x)$** instead of letting the run fail.

## Tutorial 5. The geometric programming problem from the slides
#
The *Example: Geometric programming problem* slide has 14 variables $z_i$ with 4 equality
constraints. Eliminating $z_1, z_2, z_3, z_6$ with the equalities gives the 10-variable
problem below (the same formulation is used in the OMADS test suite), with 6 inequality
constraints and bounds $10^{-6} \le x_i \le 10$:

```python
def geometric_programming(x):
    x = np.asarray(x, dtype=float)
    z3 = np.sqrt(x[3] ** 2 + x[4] ** -2 + x[5] ** -2 + x[6] ** 2)
    z6 = np.sqrt(x[6] ** 2 + x[7] ** 2 + x[8] ** 2 + x[9] ** 2)
    z1 = np.sqrt(z3 ** 2 + x[0] ** -2 + x[1] ** 2)
    z2 = np.sqrt(x[1] ** 2 + z6 ** 2 + x[2] ** 2)
    f = z1 ** 2 + z2 ** 2
    c = [z3 ** -2 + x[0] ** 2 - x[1] ** 2,
         x[1] ** 2 + z6 ** -2 - x[2] ** 2,
         x[3] ** 2 + x[4] ** 2 - x[6] ** 2,
         x[3] ** -2 + x[5] ** 2 - x[6] ** 2,
         x[6] ** 2 + x[7] ** -2 - x[8] ** 2,
         x[6] ** 2 + x[7] ** 2 - x[9] ** 2]
    return [float(f), [float(v) for v in c]]


param_gp = {"name": "GP", "baseline": [1.0] * 10, "lb": [1e-6] * 10, "ub": [10.0] * 10,
            "var_names": [f"x{i + 1}" for i in range(10)], "scaling": 10.0, "constraints_type": ["PB"] * 6}
summary_gp, hist_gp, wall_gp = solve(geometric_programming, param_gp, options={"budget": 2000},
                                     search_options={"ns": 20})
print(f"fmin = {summary_gp['fmin'][0]:.4f} after {len(hist_gp)} calls ({wall_gp:.1f} s)")
print("xmin =", np.round(summary_gp["xmin"], 4))
print("constraints at xmin (must be <= 0):", np.round(geometric_programming(summary_gp["xmin"])[1], 6))

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(np.arange(1, len(hist_gp) + 1), hist_gp["best feasible f"], color=COLORS["mads"], lw=2)
ax.set(xlabel="blackbox evaluations", ylabel="best feasible $f$", ylim=(15, 40),
       title="Geometric programming: progress of OMADS (mads, PB)")
plt.show()
```

Always **check the reported solution independently**: here we re-evaluate the constraints
at `xmin`. All six constraints are feasible and **active** (values $\approx 0$): every design
limit drives the optimum - the derivative-free counterpart of the KKT analysis from the NLP
lectures. The best value found, $f \approx 17.59$, lies within the range $15 < f \le 25$ quoted
for this problem in the OMADS test suite.

## Tutorial 6. Scaling up: 10-dimensional Rosenbrock
#
Direct search slows down as the dimension grows: each Poll costs $2n$ evaluations. With a
budget of 5000 evaluations:

```python
param_rb10 = {"name": "RB10", "baseline": [-2.5] * 10, "lb": [-5] * 10, "ub": [10] * 10,
              "var_names": [f"x{i + 1}" for i in range(10)], "scaling": [15.0] * 10}
runs_rb10 = {s: solve(rosenbrock, param_rb10, solver=s, options={"budget": 5000}, search_options={"ns": 66})
             for s in ("poll", "mads")}
pd.DataFrame([report(k, *v) for k, v in runs_rb10.items()]).drop(columns="xmin")
```

```python
fig, ax = plt.subplots(figsize=(7.5, 4.5))
for name, (_, hist, _) in runs_rb10.items():
    ax.semilogy(np.arange(1, len(hist) + 1), np.maximum(hist["best feasible f"], 1e-16),
                color=COLORS[name], lw=2, label=name)
ax.set(xlabel="blackbox evaluations", ylabel="best $f$ so far", title="10-D Rosenbrock")
ax.legend()
plt.show()
```

In 10 dimensions the Poll alone is still far from the optimum after 5000 evaluations, while
`mads` solves the problem: the Search step, which samples the space using everything evaluated
so far, matters more and more as $n$ grows.

## Takeaways
#
1. An OMADS blackbox returns `[f, [c_1, ..., c_m]]` with $c_j \le 0$ feasible; the problem and
   the algorithm are configured through the `param`, `options` and `search` dictionaries.
2. Use the `mads` solver (Search + Poll) by default; the Poll alone is slow on curved valleys
   and in higher dimensions.
3. Keep `rich_direction=True` for nonsmooth blackboxes: dense poll directions escape the kinks
   that stop coordinate search and GPS.
4. Return measurable constraints as $c_j(x)$ and choose PB (uses the violation) or EB
   (rejects infeasible points); return `inf` for crashed evaluations and never let the
   blackbox raise.
5. Always re-evaluate and check the reported solution (objective, constraints, activity).
